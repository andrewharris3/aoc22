```python
from dataclasses import dataclass, field
from collections import Counter

with open("input.txt") as f:
    commands = [y.strip() for y in f.readlines()]
```

```python
cwd = []
directory_sizes = Counter()


for line in commands:
    parts = line.split()
    match parts:
        case ("$", "cd", "/"):
            cwd = []

        case ("$", "cd", ".."):
            cwd.pop()

        case ("$", "cd", dir_name):
            cwd.append(dir_name)

        case ("dir", _):
            ...

        case (size, _) if size.isnumeric():
            size = int(size)

            for i, _ in enumerate(cwd):
                directory_sizes["/" + "/".join(cwd[:i + 1])] += size
        
```

```python
small_files = Counter({ 
        k: v for k, v in directory_sizes.items() if v <= 100_000
})
```
