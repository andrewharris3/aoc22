```python
from dataclasses import dataclass, field
from collections import Counter

with open("input.txt") as f:
    commands = [y.strip() for y in f.readlines()]
```

```python
cwd = []
directory_sizes = Counter()


for line in commands:
    parts = line.split()

    match parts:
        case ("$", "cd", "/"):
            cwd = []

        case ("$", "cd", ".."):
            cwd.pop()

        case ("$", "cd", dir_name):
            cwd.append(dir_name)

        case ("dir", _):
            ...

        case (size, _) if size.isnumeric():
            size = int(size)

            for i in range(len(cwd) + 1):
                path = "/" + "/".join(cwd[:i])
                directory_sizes[path] += size
```

```python
small_files = Counter({ 
        k: v for k, v in directory_sizes.items() if v <= 100_000
})
```

```python
total_disk_space = int(7e7)
used = directory_sizes["/"]
free = total_disk_space - used
required_space = int(3e7)

print(f"""
    {total_disk_space=:,.0f}, 
    {used=:,.0f}, 
    {free=:,.0f},
    {required_space=:,.0f}
""")

disk_space_to_free = required_space - free
```

```python
min((v for _, v in directory_sizes.items() if v >= disk_space_to_free))
```
