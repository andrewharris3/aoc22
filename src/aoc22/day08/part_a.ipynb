```python
from dataclasses import dataclass, field
import numpy as np

with open("input.txt") as f:
    trees = [y.strip() for y in f.readlines()]

(trees_parsed := np.array([[int(i) for i in x] for x in trees]))
```

```python
(edge_trees := trees_parsed.shape[0] * 2 + trees_parsed.shape[1] * 2 - 4)
```

```python
left_scan = np.maximum.accumulate(trees_parsed, axis=1)
top_scan = np.maximum.accumulate(trees_parsed, axis=0)
bottom_scan = np.maximum.accumulate(trees_parsed[::-1, :], axis=0)[::-1, :]
right_scan = np.maximum.accumulate(trees_parsed[:, ::-1], axis=1)[:, ::-1]
```

```python
left_shift = np.pad(left_scan, ((0, 0), (1, 0)), constant_values  = -1)[:, :-1] # Pads to the left and removes last column
right_shift = np.pad(right_scan, ((0, 0), (0, 1)), constant_values  = -1)[:, 1:] # Pads to the right and removes the first column
top_shift = np.pad(top_scan, ((1, 0), (0, 0)), constant_values  = -1)[:-1, :] # Pads to the top and removes the last row
bottom_shift = np.pad(bottom_scan, ((0, 1), (0, 0)), constant_values  = -1)[1:, :] # Pads to the bottom and removes the first row
```

```python
visible_left = trees_parsed > left_shift
visible_right = trees_parsed > right_shift
visible_top = trees_parsed > top_shift
visible_bottom = trees_parsed > bottom_shift
```

```python
visible = visible_left | visible_right | visible_top | visible_bottom
```
